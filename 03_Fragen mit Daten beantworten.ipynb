## Vergleiche zwischen den Genres
Wir haben bereits gesehen, wie wir zeitliche Trends analysieren können. Eine weitere Unterscheidung der Lieder kann nach dem Genre vorgenommen werden. Es stellt sich die Frage, ob sich Lieder verschiedener Genres unterschiedliche Werte in den Merkmalen haben.

In der nächsten Zelle werden Daten aus unserer Tabelle selektiert. Dies geschieht so, dass in der Spalte *genre* nach den Wörtern *pop* oder *rock* gesucht wird und nur die Zeilen, in denen das zutrifft, selektiert werden. Von diesen Teilbereichen werden anschließend jeweils Mittelwerte einer Spalte ausgerechnet und ausgegeben. 

<font color='blue'>**Aufgabe:**</font> Finde heraus, ob sich Lieder dieser beiden Genres in einem Merkmal (duration_ms, danceability, energy, loudness, tempo, ...) unterscheiden. Lieder welches Genre sind beispielsweise lauter, schneller, energischer, ...? Notiere deine Ergebnisse in der übernächsten Zelle.

````python
pop_songs = liederdaten[liederdaten['genre'].str.contains('pop')]
rock_songs = liederdaten[liederdaten['genre'].str.contains('rock')]

merkmal = 'danceability'

print("Mittelwert für Pop:", pop_songs[merkmal].mean().round(3))
print("Mittelwert für Rock:", rock_songs[merkmal].mean().round(3))
````

Wir können uns die Namen der Artists und die jeweilige Liederanzahl in einer neuen Tabelle speichern. Dies wird in der nächsten Zelle gemacht. Dazu wurden bereits die Mittelwerte für das Tempo berechnet. Das erlaubt uns anschließend zu analysieren, ob es Muster bei den Artists gibt, die schon häufiger Lieder veröffentlicht haben. 

````python
artists = pd.DataFrame(liederdaten['artist'].unique(), columns=['Artist'])
artists['Liederanzahl'] = liederdaten['artist'].value_counts().sort_index(ascending=True).values
artists['Mittelwert'] = liederdaten.groupby('artist')['tempo'].mean().sort_index(ascending=True).values
artists
````
