## Diagramme erstellen

**Ziel:** Damit wir einen besseren Überblick über die Daten in der Spalte energy bekommen, erstellen wir nun ein Diagramm. 

````python
liederdaten['popularity'].plot(kind="box", title="Boxplot für die popularity-Werte")
````

**Aufgabe:** Was sagt das Diagramm aus? Interpretiere das Diagramm in der nächsten Zelle schriftlich.

Um einen Überblick über die Verteilung der Werte zu sehen, könnte auch ein Säulendiagramm hilfreich sein. Aktuell gibt es aber zu viele verschiedene Werte in der Spalte *popularity*. Daher nutzen wir ein sogenanntes Histogram. Dort werden die Lieder in einige Intervalle aufgeteilt. 

````python
liederdaten['popularity'].plot(kind="hist", bins=3, 
                           xlabel="popularity-Werte", ylabel="Anzahl Lieder", 
                           title="Häufigkeit der popularity-Werte unterteilt in Intervalle")
````
